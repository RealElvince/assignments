```python
import numpy as np
from scipy.linalg import lu
```

```python
# define matrix A

A = np.array([[4,3],[6,3]])
```

```python
# Perform LU decomposition

P, L, U = lu(A)
```

```python
# Print results
print("Matrix:")
print(A)
print("\nLower triangular matrix L:")
print(L)
print("\nUpper triangular matrix U:")
print(U)
```
