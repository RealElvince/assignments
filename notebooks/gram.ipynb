```python
import numpy as np

# Given set of linearly independent vectors
vectors = np.array([
    [1, 1, 0],
    [1, 0, 1],
    [0, 1, 1]
], dtype=float)

def gram_schmidt(vectors):
    """Perform Gram-Schmidt orthonormalization."""
    ortho_basis = []
    for v in vectors:
        w = v.copy()
        for u in ortho_basis:
            # Subtract the projection of v onto u
            w -= np.dot(u, v) * u
        # Normalize w
        w_norm = np.linalg.norm(w)
        if w_norm > 1e-10:
            u = w / w_norm
            ortho_basis.append(u)
    return np.array(ortho_basis)

# Compute the orthonormal basis
orthonormal_basis = gram_schmidt(vectors)

# Display the result
print("Orthonormal basis:")
for vec in orthonormal_basis:
    print(vec)
```
